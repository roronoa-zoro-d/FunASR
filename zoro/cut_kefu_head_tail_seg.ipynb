```python
import sys
from funasr import AutoModel
import soundfile



vad_model = AutoModel(model="fsmn-vad", model_revision="v2.0.4")
asr_model = AutoModel(model="paraformer-zh", model_revision="v2.0.4",)
```

```python


wav_scp_file = '/data/nas/dataset/asr/kefu/shidian/wav.scp'

utts = []
wav2path = {}
with open(wav_scp_file, 'r') as f:
    for line in f:
        utt, path = line.strip().split()
        utts.append(utt)
        wav2path[utt] = path
print(f'read {len(utts)} utts')


datas = []

for i, utt in enumerate(utts):
    wav_file = wav2path[utt]
    speech, sample_rate = soundfile.read(wav_file)
    num_point_ms = int(sample_rate/1000)
    
    res_vad = vad_model.generate(input=wav_file)
    segments = res_vad[0]['value']
    first_chunk = speech[num_point_ms*segments[0][0]:num_point_ms*segments[0][1]]
    final_chunk = speech[num_point_ms*segments[-1][0]:num_point_ms*segments[-1][1]]
    first_chunk_asr_res = asr_model.generate(input=first_chunk)
    first_chunk_asr = first_chunk_asr_res[0]['text']
    final_chunk_asr_res = asr_model.generate(input=final_chunk)
    final_chunk_asr = final_chunk_asr_res[0]['text']
    
    print("utt {} | {} : {} ; {} : {}".format(utt, segments[0], first_chunk_asr, segments[-1], final_chunk_asr))
    
    data = {
        "utt": utt,
        "first_chunk": {
            "vad_seg": segments[0],
            "asr_txt": first_chunk_asr,
        },
        "final_chunk": {
            "vad_seg": segments[-1],
            "asr_txt": final_chunk_asr,
        },
    }
    datas.append(data)
    
    
    if i == 10:
        break
```

```python
for data in datas:
    print(data)
```

# 按照asr_serve client_batch 的格式 挑选第一句和最后一句

```python
import json
import soundfile as sf


def read_jsonl(filename):
    datas = []
    with open(filename, 'r') as f:
        for line in f:
            datas.append(json.loads(line))
        
            if len(datas) > 100:
                break
    
    return datas


def select_first_sentence(text):
    ref_text = "你好 我是 圆通 客服， 请问 有什么 可以 帮您 很 高兴 为你 服务"
    
    if len(text) < 5:
        return False
    
    num_match = 0
    for wd in ref_text.split():
        if wd in text:
            num_match += 1

    if num_match >= 2:
        return True
    
    return False
    
def select_final_sentence(text):
    ref_text = "感谢 来电 祝 您 生活 愉快 那 不 打扰您 再见"
    
    for wd in ref_text.split():
        if wd in text:
            return True
    
    return False

    
```

# 根据已经选择好的伪标签和切块音频   选首尾句

```python
import glob 
from collections import defaultdict

segment_dir = '/data/nas/dataset/asr/kefu/huaian2/segment_wav_pl/'
out_dir = '/data/nas/dataset/asr/kefu/huaian2/segments_first_and_final/'

text_files = glob.glob(segment_dir + '*.text')

utt2segs = defaultdict(list)
for i, text_file in enumerate(text_files):
    wav_file = text_file[:-5] + ".wav"
    text_name = text_file.split('/')[-1]
    utt, suffix = text_name.split('_seg')
    idx = int(suffix.split('.')[0])
    
    text = ""
    with open(text_file, 'r') as f:
        text = f.readline().strip()
    utt2segs[utt].append([idx, text, utt, text_file, wav_file])
    
    # if i > 500:
    #     break
    
# 排序
for utt in utt2segs:
    segs = utt2segs[utt]
    segs.sort(key=lambda x: x[0])
    
    # for a in utt2segs[utt]:
    #     print(a)
    
    
    
#  选择
first_datas = []
final_datas = []
for utt in utt2segs:
    segs = utt2segs[utt]
    first_seg = segs[0]
    final_seg = segs[-1]
    
    first_seg_text = first_seg[1]
    final_seg_text = final_seg[1]
    
    if select_first_sentence(first_seg_text):
        first_datas.append(first_seg)
        
    if select_final_sentence(final_seg_text):
        final_datas.append(final_seg)
        # print(f'final text: {final_seg_text}')
        
# print(f'first text')
# for data in first_datas:
#     print(data[1])
# print(f'final text')
# for data in final_datas:
#     print(data[1])
    
with open(f'{out_dir}/select_first_seg_utt.info', 'w') as f:
    for data in first_datas:
        f.write(f'{data[2]} {data[3]} {data[4]}\n')
with open(f'{out_dir}/select_final_seg_utt.info', 'w') as f:
    for data in final_datas:
        f.write(f'{data[2]} {data[3]} {data[4]}\n')
    
```
